```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('news.csv', index_col=None)
```

```python
df
```

```python
dataset=df.drop("Unnamed: 0",axis=1)
```

```python
dataset
```

```python
y=dataset["label"]
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(dataset['text'], y, test_size=0.33, random_state=53)
```

```python
hashing_vectorizer =HashingVectorizer(stop_words='english')
hashing_train = hashing_vectorizer.fit_transform(X_train)
print(hashing_train)
hashing_test = hashing_vectorizer.transform(X_test)
```

```python
from sklearn.feature_extraction.text import HashingVectorizer

hashing_vectorizer = HashingVectorizer(n_features=2**20)  
print("Number of features:", hashing_vectorizer.n_features)
```

```python
from sklearn.feature_extraction.text import HashingVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics

hashing_vectorizer = HashingVectorizer(n_features=2**20, alternate_sign=False)
hashing_train = hashing_vectorizer.transform(X_train)
hashing_test = hashing_vectorizer.transform(X_test)

clf = MultinomialNB()
clf.fit(hashing_train, y_train)
pred = clf.predict(hashing_test)

score = metrics.accuracy_score(y_test, pred)
print("accuracy: %0.3f" % score)
```

```python
from sklearn.metrics import classification_report

report=classification_report(y_test, pred)
```
